```python
import ipywidgets as widgets
from IPython.display import display

```

```python
text = widgets.Text(description='Name')
display(text)
```

```python
button = widgets.Button(description='Click Me!')
display(button)

```

```python
rbutton = widgets.RadioButtons(options=['male','female'],description='gender -->')
display(rbutton)
```

```python
x=0
clk = widgets.Button(description='Click Me!')
out = widgets.Output()

def click(b):
    global x
    while True:
      with out:
          x+=1
          print(x , "Speed Breaker @!#$")
          if x==10:
            break

clk.on_click(click)
display(clk, out)
```

```python
slider = widgets.IntSlider(description='slider',min=1,max=100,step=1.5,readout=True)
display(slider)
```

```python
weight = widgets.FloatText(
    description='Weight (kg):'
)

height = widgets.FloatText(
    description='Height (m):'
)

button = widgets.Button(
    description='Calculate BMI',
    button_style='success'
)

output = widgets.Output()

def calculate_bmi(b):
    with output:
        output.clear_output()

        if height.value <= 0:
            print("Please enter a valid height.")
            return

        bmi = weight.value / (height.value ** 2)

        print(f"BMI = {bmi:.2f}")

        if bmi < 18.5:
            print("Category: Underweight")
        elif bmi < 25:
            print("Category: Normal Weight")
        elif bmi < 30:
            print("Category: Overweight")
        else:
            print("Category: Obese")

button.on_click(calculate_bmi)

display(weight, height, button, output)
```
