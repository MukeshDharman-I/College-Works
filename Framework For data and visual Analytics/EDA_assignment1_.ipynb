```python
import pandas as pd
```

```python
df = pd.read_csv("students_missing_data.csv")
```

```python
print(df.head())
print("_______________________")
print(df.describe())
print("_______________________")
```

```python


print(df['Gender'])
print("_______________________")
print("_______________________")
for i in range(len(df['Gender'])):
  if df['Gender'][i]=='Not Disclosed' and df['Gender'][i-1]=='M':
    #df['Gender'][i]='F'
    df.loc[i, "col"] = 'F'
  elif df['Gender'][i]=='Not Disclosed' and df['Gender'][i-1]=='F':
    #df['Gender'][i]='M'
    df.loc[i, "col"] = 'M'
  print(df['Gender'][i])
print("_______________________")
print("_______________________")
print(df['Gender'])
```

```python
df['Age'].fillna(df['Age'].median(),inplace=True)
df['Age']=df['Age'].astype(int)
print(df['Age'])
```

```python
df.dropna(subset=['ExamScore'], inplace=True)
print(df['ExamScore'])
```

```python
df.dropna(subset=['AttendancePercent'],inplace=True)
print(df['AttendancePercent'])
```

```python
df=df.drop(columns=['StudentID'],axis=1)

print(df)
```
